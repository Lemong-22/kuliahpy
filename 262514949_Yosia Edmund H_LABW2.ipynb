## A1: Konversi Suhu

Seorang perawat mencatat suhu tubuh pasien sebesar **37.5 derajat Celsius**.

Buat program yang:
1. Menyimpan suhu Celsius ke dalam sebuah variabel (gunakan tipe `float`).
2. Mengubahnya ke **Fahrenheit** dengan rumus: `F = C * 9 / 5 + 32`
3. Mengubahnya ke **Kelvin** dengan rumus: `K = C + 273.15`
4. Menampilkan ketiga hasilnya.

### Output yang Diharapkan
```
Suhu dalam Celsius : 37.5 C
Suhu dalam Fahrenheit : 99.5 F
Suhu dalam Kelvin : 310.65 K
```

## A2: Pembuat Username Otomatis

Sebuah sistem kampus membuat username mahasiswa secara otomatis.

Diketahui data mahasiswa:
- Nama depan : `Ethan`
- Nama belakang : `Lim`
- Tahun lahir : `2005` (simpan sebagai **integer**)

Buatlah program yang dapat:
1. Menyimpan ketiga data di atas ke dalam variabel.
2. Membuat `nama_lengkap` dengan menggabungkan nama depan dan nama belakang
   menggunakan **concatenation (`+`)**, dipisah satu spasi.
3. Membuat `username` dengan format: **nama depan + garis bawah + tahun lahir**.
4. Menampilkan nama lengkap, panjangnya, username, dan panjang username.

### Output yang Diharapkan
```
Nama lengkap : Michelle Tanuwijaya
Panjang nama : 19 karakter
Username : Michelle_2005
Panjang username : 13 karakter
```

````python
Nama_depan = "Ethan"
Nama_belakang = "Lim"
Tahun_lahir = 2005

Nama_lengkap = Nama_depan + " " + Nama_belakang
panjang_nama = len(Nama_lengkap)
username = Nama_depan + "_" + str(Tahun_lahir)
panjang_username = len(username)

print(f'Nama lengkap : {Nama_lengkap}\nPanjang nama : {panjang_nama} karakter\nUsername : {username}\nPanjang username : {panjang_username} karakter')
````
