## Soal 1 : Perkenalan Diri

Buatlah suatu program yang dapat:
1. Meminta pengguna menginputkan **nama**, **umur** (integer), dan **tinggi badan dalam meter** (float).
2. Menampilkan kalimat perkenalan.
3. Menampilkan umur pengguna **10 tahun lagi** (perhatikan contoh).

### Contoh Output
```
input your name: Andi --> input
input your age: 19 --> input
input your body height (in meter): 1.72 --> input
```

````python
# masukkan nama bertipe string
nama = input("input your name: ")
#masukkan umur bertipe integer
umur = int(input("input your age: "))
#masukkan tinggi badan bertipe float
tinggi = float(input("input your body height (in meter): "))
#menampilkan output
print(f"Halo, nama saya {nama}\nUmur saya {umur} tahun dan tinggi saya {tinggi} m")
print(f"10 tahun lagi umur saya {umur+10} tahun")
````

## Soal 2 : Menghitung Panjang Nama

Buatlah program sederhana yang dapat:
1. Meminta pengguna memasukkan **nama depan** dan **nama belakang**.
2. Menggabungkan keduanya menjadi satu variabel `nama_lengkap` menggunakan **concatenation (`+`)**,
   dengan satu spasi di antaranya.
3. Menampilkan `nama_lengkap` beserta jumlah karakternya menggunakan `len()`.
4. Menampilkan juga jumlah karakter nama lengkap **tanpa menghitung spasi**.

### Contoh Output
```
Nama depan: Kelly --> input
Nama belakang: Kirsten --> input

Nama lengkap: Kelly Kirsten
Jumlah karakter (dengan spasi): 13
Jumlah karakter (tanpa spasi): 12
```

````python
# masukkan nama depan dan belakang
nama_depan = input("Nama depan: ")
nama_belakang = input("Nama belakang: ")
#gabungkan menjadi nama lengkap
nama_lengkap = nama_depan + " " + nama_belakang
#menghitung jumlah karakter dengan spasi
jumlah_char1 = len(nama_lengkap)
#menghitung jumlah karakter tanpa spasi
nama_lengkap_nospace = nama_depan+nama_belakang
jumlah_char2 = len(nama_lengkap_nospace)

#menampilkan output
print(f"Nama lengkap: {nama_lengkap}")
print(f"Jumlah karakter (dengan spasi): {jumlah_char1}")
print(f"Jumlah karakter (tanpa spasi): {jumlah_char2}")
````

````python
#meminta nama barang (string), jumlah barang (integer), dan harga satuan (float).
nama_barang = input("Nama barang: ")
jumlah_barang = int(input("Jumlah barang: "))
harga_satuan = float(input("Harga satuan: "))

#menghitung subtotal = jumlah * harga
subtotal = jumlah_barang * harga_satuan
#menghitung PPN sebesar 11% dari subtotal
ppn = subtotal * 0.11

#menghitung total = subtotal + ppn
total = subtotal + ppn 
#menampilkan output
print("----- STRUK BELANJA -----")
print(f"Barang : {nama_barang}\nJumlah : {jumlah_barang}\nSubtotal : {subtotal}\nPPN (11%) : {ppn}\nTotal : {total}")
````

## Soal 4 : Ongkos Potong Rumput

Sebuah taman berbentuk persegi panjang mengelilingi sebuah rumah (lihat ilustrasi di slide 29).

Silahkan copy-paste code yang telah Anda buat di sini!

````python
#define panjang dan lebar taman, rumah
panjang_taman = 12
lebar_taman = 10
panjang_rumah = 8
lebar_rumah = 5
#hitung luas taman dan rumah
luas_taman = panjang_taman * lebar_taman
luas_rumah = panjang_rumah * lebar_rumah
#hitung luas rumput
luas_rumput = luas_taman - luas_rumah
#hitung biaya total
total_biaya=luas_rumput*1000
#menampilkan output
print(f'total biaya potong rumput adalah Rp {total_biaya}')
````

## Soal 6 : Kutipan dan Tanda Kutip

Buat program yang:
1. Meminta pengguna menginputkan **nama tokoh** dan sebuah **quote**.
2. Menampilkan quote dengan **diapit tanda kutip ganda (`"`)**, diikuti nama tokohnya.
3. Menampilkan panjang quote (jumlah karakter) menggunakan `len()`.

Perhatikan contoh di bawah

### Contoh Output
```
Nama tokoh: Ada Lovelace --> input
Kutipan: The Analytical Engine has no pretensions whatever to originate anything. It can do whatever we know how to order it to perform. --> input

"The Analytical Engine has no pretensions whatever to originate anything. It can do whatever we know how to order it to perform." - Ada Lovelace
quote length: 36 karakter
```

````python
# input nama tokoh dan quote nya
nama_tokoh = input("Nama tokoh: ")
kutipan = input("Kutipan: ")
#menghitung panjang quote
panjang_quote = len(kutipan)
#menampilkan output
print(f'"{kutipan}" - {nama_tokoh}')
print(f"quote length: {panjang_quote} karakter")
````
